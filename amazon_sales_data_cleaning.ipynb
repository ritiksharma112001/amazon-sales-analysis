### Converting DataType of date column

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df[df["Amount"].isna()]["Status"].value_counts()
```

```python
df.isna().sum()
```

```python
df["Courier Status"] = df["Courier Status"].fillna("Unknown")
df["currency"] = df["currency"].fillna("INR")
df["ship-city"] = df["ship-city"].fillna("Unknown")
df["ship-state"] = df["ship-state"].fillna("Unknown")
df["ship-postal-code"] = df["ship-postal-code"].fillna("NA")
df["ship-country"] = df["ship-country"].fillna("Unknown")
df["promotion-ids"] = df["promotion-ids"].fillna("No promotion")
df["fulfilled-by"] = df["fulfilled-by"].fillna("NA")
df.drop(columns=["Unnamed: 22"], inplace=True)
```

```python
df.isna().sum()
```

```python
df.loc[(df["Status"] == "Cancelled") & (df["Amount"].isna()), "Amount"] = 0
```
